#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
basketball <- read_csv("https://raw.githubusercontent.com/diverse-data-hub/diversedata/refs/heads/main/data-clean/womensmarchmadness.csv") |>
                select(school, conf_wins, reg_wins, tourney_finish)
#basketball
summary <- basketball |>
            drop_na(conf_wins, reg_wins) |>
            #summarize()
            mutate(mean_conf_wins = mean(conf_wins)) |>
            mutate(mean_reg_wins = mean(reg_wins)) |>
            mutate(min_conf_wins = min(conf_wins)) |>
            mutate(min_reg_wins = min(reg_wins)) |>
            mutate(max_conf_wins = max(conf_wins)) |>
            mutate(max_reg_wins = max(reg_wins)) |>
            select(mean_conf_wins, mean_reg_wins, min_conf_wins, min_reg_wins, max_conf_wins, max_reg_wins) |>
            slice(1)
summary
# make this more concise with summarize()
```

#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).
- Show the outputted table. 
</div>

```r
basketball_tidy <- basketball |>
                    drop_na(conf_wins, reg_wins)
#basketball_tidy
no_of_na <- basketball |>
            mutate(across(everything(), as_factor)) |>
            pivot_longer(
                cols = everything(),
                names_to = "all_columns",
                values_to = "values") |>
            group_by(all_columns) |>
            summarize(missing_count = sum(is.na(values)))
no_of_na
             #summarize(across(everything(), function(x) sum(is.na(x))))
             #summarize(across(everything(), ~ sum(is.na(.))))
# how can I do this without using ~ and is.na(.)? what is (.)?
```
